```python
import pandas as pd

# Membaca file yang tadi di-upload
dataraw = pd.read_excel("Alichia.xlsx")

print("Tabel Data")
display(dataraw)
```

```python
#Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print(datafrq)
```

```python
#Grafik Grais
datafrq.plot()
```

```python
#Grafik Batang
datafrq.plot(kind='bar')
```

```python
# Membuat tabel frekuensi Grade
datafqr = dataraw["Grade"].value_counts().reset_index()

# Mengubah nama kolom
datafqr.columns = ["Grade", "Frekuensi"]

display(datafqr)
```

```python
#Pie Chart
datafqr.plot(kind='pie',y='Frekuensi',autopct='%1.0f%%',legend=False)
```

```python
# Statistika Deskriptif

dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors="coerce")

dt = dataraw["Final Score"]

stats = dt.describe()

stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["Variance"] = dt.var()
stats["Range"] = dt.max() - dt.min()
stats["Skewness"] = dt.skew()
stats["Kurtosis"] = dt.kurt()

print(stats)
```
